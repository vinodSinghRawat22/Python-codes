```python
import numpy as np
```

```python
NUMPY INDEXING AND SLICING OF VECTORS 
```

```python
arr = np.arange(11,21)
arr
```

```python
arr[6]
```

```python
slice = arr[1:5]

slice[1] = 3
```

```python
arr
```

```python
NUMPY INDEXING AND SLICING OF MATRIX 
```

```python
arr = np.arange(10, 40).reshape(6,5)
arr
```

```python
arr[4]  # row slicing  
```

```python
arr[:,2 ]    # column slicing 
```

```python
arr[2,1]   # element slicing from matrix 
```

```python
arr[0:2, 1:3]   # sub-matrix slicing from matrix 
```

```python
slice = arr[0:2, 1:3]
slice[0,0] = 5                       #if we change any part of sliced matrix it will also reflect to main matrix 
```

```python
arr
```

```python
arr[3:6, 3:5]   # sub-matrix slicing from main matrix 
```

```python
BOOLEAN INDEXING 
```

```python
arr = np.arange(10, 20)
arr
```

```python
bool_index = arr %2 == 0
bool_index
```

```python
arr = arr[bool_index]
arr
```
