```python
import numpy as np
```

```python
import pandas as pd
```

```python
labels= ['a', 'b', 'c']
my_list = [1,2,3]
arr = np.array([4,5,6])
my_dict = {'science': 10, 'Maths': 20, 'computer':30}
```

```python
# we can create series using list, array, dictonery etc. it contain index numbers in lable  if you did not provided labels.
```

```python
pd.Series(my_list)   # series using list and the first column is index numbers 
```

```python
pd.Series(arr)  # series using array and the first column is index numbers 
```

```python
pd.Series(my_dict)   # series using dictonery and the first column is keys 
```

```python
pd.Series(arr, index=labels)  # here we using arr and using labels from labels list 
```
