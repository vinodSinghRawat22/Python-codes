```python
import numpy as np 
import pandas as pd 
```

**PIVOT TABLES**

```python

data = {
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South',
               'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South',
               'East', 'West', 'North', 'South'],
    'Sales': [538, 847, 732, 681, 912, 650, 789, 723, 810, 905, 600, 710, 
              830, 760, 780, 695, 875, 720, 755, 690],  # Same count as image (20)
    'Units': [45, 65, 58, 52, 60, 49, 55, 50, 62, 70, 48, 57, 
              59, 54, 53, 51, 66, 56, 58, 52],
    'Rep': ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 
            'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 
            'John', 'Mary', 'Bob', 'Alice'],
    'Date': pd.date_range('2023-01-01', periods=20)
}

df = pd.DataFrame(data)

df['Month'] = df['Date'].dt.month_name()
df['Quarter'] = 'Q' + df['Date'].dt.quarter.astype(str)

df
```

```python
# Pivot table is used for Aggrigation purpose 
```

```python
# Creates a pivot table showing Sales for each Region and Product combination
pd.pivot_table(df, values="Sales", index='Region', columns='Product')
```

```python
# You can change the aggregation method using the aggfunc parameter, e.g. sum(), median(), etc. and by default it is set to mean
pd.pivot_table(df, values='Sales', index='Region', columns='Product', aggfunc='median')  
# aggfunc='sum' will calculate total sales instead of the default average
```

```python
pd.pivot_table(df, values=['Sales', 'Units'], index='Region', columns='Product')  
```
