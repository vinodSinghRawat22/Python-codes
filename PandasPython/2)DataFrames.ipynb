```python
#creating a data frame==================================
```

```python
data = {
    "Name": ["John", "Michel", "Sara", "Liam"],
    "City": ["New York", "Paris", "London", "Toronto"],
    "Age": [25, 30, 22, 28],
    "Occupation": ["Engineer", "Designer", "Doctor", "Developer"]
}
df1 = pd.DataFrame(data)
df1
```

```python
data_list = [
    ["John", 25, "New York", "Engineer"],
    ["Michel", 30, "Paris", "Designer"],
    ["Sara", 22, "London", "Doctor"],
    ["Liam", 28, "Toronto", "Developer"]
]
column = ["Name", "Age", "City", "Occupation"]
df2 = pd.DataFrame(data_list, columns= column, index=range(1,len(data_list)+1))
df2
```
