```python
import numpy as np 
import pandas as pd
```

```python
# loading anime.csv in this notebook 
```

```python
df = pd.read_csv(r'anime.csv')
```

```python
df.head()
# it will show top 5 rows by default and we can access more by giving a attribute n = number of rows 
```

```python
# Make a new column for episode count
```

```python
def episode_extract(txt):
    check = False
    data = ""
    for i in txt : 
        if i == ')':
            check = False
            break 
        if check == True:
            data += i
        if i == '(':
            check = True
         

    return data 

    
```

```python
df['Episodes'] = df['Title'].apply(episode_extract)
df['Episodes'] = df['Episodes'].str.replace("eps", "")
df['Episodes'] = df['Episodes'].astype(int)
df
```

```python
# Make a new column for time stamp 
```

```python
def extract_timeStamp(txt):
    data = ""
    for i in range(len(txt)): 
        if txt[i] == ')':
            for j in range(i+1, i+20):
                data+=txt[j]
    return data 
```

```python
df['Total time'] = df['Title'].apply(extract_timeStamp)
df.head()
```

```python
#Make a new column of  total months: 
def calculate_months(time_period):
    start, end = time_period.split(' - ')
    start_date = pd.to_datetime(start)
    end_date = pd.to_datetime(end)
    return (end_date.year - start_date.year) * 12 + (end_date.month - start_date.month) + 1

df['Total Months'] = df['Total time'].apply(calculate_months)
df.head()
```

```python
# Which anime has highest score  
df[df['Score'] == df['Score'].max()]['Title']
```

```python
# give me top 5 highest scoring anime 
df.nlargest(5, 'Score')
```

```python
# Which anime has the highest episode count 
df[df['Episodes']== df['Episodes'].max()]
```

```python
# animes with top 5 episode count 
df.nlargest(5, 'Episodes')[['Title', 'Episodes']]
```

```python
# which is the longest running anime 
df[df['Total Months'] == df['Total Months'].max()][['Title', 'Total Months']]
```
