```python
import numpy as np

# Example dictionary for missing data concept
data = {
    "Name": [ "Bhavya",np.nan, "Chirag", np.nan, "Divya", "Esha", np.nan],
    "Age": [18, 19, np.nan, 20, np.nan, 21, 22],
    "Marks": [85, np.nan, 90, 88, np.nan, 92, 80],
    "City": ["Delhi", np.nan, "Mumbai", np.nan, "Bangalore", np.nan, "Kolkata"]
}
df = pd.DataFrame(data)
df
```

```python
df.isna()  # It checks the none values and returns true where none value exists 
```

```python
df.isna().sum()  #it counts the number of null values  column wise
```

```python
df.isna().any()  # this returns flase if any column has no null values 
```

```python
#Removing null values  (it works on the basis of rows )
```

```python
df
```

```python
df.dropna()  # it will remove all the rows except the rows which doesn't have null value in it
```

```python
df
```

```python
df.dropna(thresh=3) # it will remove the rows which doesn't have minimum 3 real values 
```

```python
#Filling the missing data 
```

```python
df
```

```python
df.fillna(0)  # This will fill the null value places with a given value (column wise selection )
```

```python
values = {'Name':'xyz', 'Age':'19', 'Marks': 33, 'City': 'Delhi' }
```

```python
df.fillna(value= values)# This is to fill a particular column with a specific value (column wise selection )
```
