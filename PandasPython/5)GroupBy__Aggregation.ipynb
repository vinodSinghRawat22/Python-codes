```python
import numpy as np 
import pandas as pd 
```

**GROUP BY**

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store':    ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales':    [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date':     pd.date_range('2023-01-01', periods=8)
}

df = pd.DataFrame(data)
```

```python
df
```

```python
# Group by Category and calculate the sum of sales 
```

```python
cat = df.groupby('Category')['Sales'].sum()
cat 
```

```python
# Group by Store and calculate the sum of sales 
```

```python
cat = df.groupby('Store')['Sales'].sum()
cat
```

```python
#Group by multiple culumns 
#Group by Catogery and Store 
```

```python
cat = df.groupby(['Category','Store'])['Sales'].sum()
cat
```

**AGGREGATION**

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store':    ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales':    [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date':     pd.date_range('2023-01-01', periods=8)
}

df = pd.DataFrame(data)
```

```python
df
```

```python
df['Sales'].mean()
```

```python
#Other functions like : median, min, max, count, std
```

```python
df['Sales'].std()
```

```python
df['Sales'].agg(['mean', 'median', 'min', 'max', 'count', 'std'])
```

```python
df['Sales'].mode()
```
