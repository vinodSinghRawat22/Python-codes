```python
import numpy as np
import pandas as pd
```

**MERGING TWO DATA FRAMES**

```python
df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [24, 27, 22]
})

df2 = pd.DataFrame({
    'ID': [1, 2, 4],
    'City': ['New York', 'Los Angeles', 'Chicago'],
    'Country': ['USA', 'USA', 'USA']
})

df1
```

```python
df2
```

```python
pd.merge(df1,df2, on = 'ID', how= 'inner')
# 'on' defines the common column used to match rows (ID in this case)
# 'how="inner"' returns only the rows where IDs are present in both DataFrames
```

```python
pd.merge(df1,df2, on = 'ID', how= 'outer')
# 'on' specifies the common column used for merging (ID here)
# 'how="outer"' returns all rows from both DataFrames, matching where possible and filling missing values with NaN
```

```python
# 'on' tells which common column to use for merging (ID here)
# 'how="left"' keeps all rows from df1 an
```

```python
pd.merge(df1,df2, on = 'ID', how= 'right') 
# 'on' specifies the common column used for joining (here it's 'ID')
# 'how="right"' keeps all rows from df2 and only matching rows from df1
```

**CONCATINATION OF TWO DATA FRAMES**

```python
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2', 'A3'],
    'B': ['B0', 'B1', 'B2', 'B3'],
    'C': ['C0', 'C1', 'C2', 'C3']
})

df2 = pd.DataFrame({
    'A': ['A4', 'A5', 'A6', 'A7'],
    'B': ['B4', 'B5', 'B6', 'B7'],
    'C': ['C4', 'C5', 'C6', 'C7']
})
df1
```

```python
df2
```

```python
pd.concat([df1, df2]) 
# This stacks df2 below df1 (row-wise) without changing original DataFrames
```

```python
pd.concat([df2, df1]) 
# This stacks df1 below df2 (row-wise) without changing original DataFrames
```

```python
pd.concat([df1, df2], axis = 1)  
# Joins df1 and df2 side-by-side (column-wise), matching rows by index
```

**JOINING OF 2 DATA FRAMES**

```python
df1 = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David']
}, index = [1, 2, 3, 4])
df2 = pd.DataFrame({
    'Score': [80, 85, 90, 95]
}, index = [ 2, 3, 4, 5])
df1
```

```python
df2
```

```python
df1.join(df2) 
# inner join which joins df2 in df1 and the df2 does not conatain 1 so in place of score of 1st one there is NaN is produced as output 
# it contain only df1 items and the same id items from df2 table like here from score data frame the 5 is not joined 
# and here i written df1 first so the df1's column is printed first 
```

```python
df2.join(df1)
#here 1 is not joined because i joining df1 in df2
```

```python
df1.join(df2, how= 'outer') 
# Here both the data frames items are merged and in 1st one's score and 5th ones name there is NaN 
```
