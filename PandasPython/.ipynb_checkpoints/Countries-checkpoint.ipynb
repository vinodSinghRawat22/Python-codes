```python
# which country has the highest population 


df[df['population'] == df['population'].max()][['country', 'population']]
```

```python
# what is the capital of the country with highest population 


df[df['population'] == df['population'].max()][['country', 'capital_city']]
```

```python
# which country has the highest population


df[df['population'] == df['population'].min()][['country', 'population']]
```

```python
# what is the capital of the country with least population 


df[df['population'] == df['population'].min()][['country', 'capital_city']]
```

```python
# Give me top five countries with highest democratic score 


# df.sort_values(by = 'democracy_score', ascending= False) ['country'].head()
# or
df.nlargest(5, 'democracy_score')[['country', 'democracy_score']]
```

```python
# How many total regions are there 


# x = df['region'].unique()
# len(x)
#or
int(df['region'].value_counts().count())
```

```python
# How many countries lie in Eastern Europe region 


int(df[df['region'] == 'Eastern Europe'].count()['country'])
# or
# int(df[df['region'] == 'Eastern Europe'].shape[0])
```

```python
df[df['region'] == 'Eastern Europe']['country']
```

```python
# Who is the political leader of 2nd highest populated country 


# df.nlargest(2,'population').iloc[1][['country','political_leader']]
# or

# df.sort_values(by = 'population', ascending= False)[['country', 'political_leader']].iloc[1]
# or

df[df['population'] == df['population'].nlargest(2).iloc[1]][['country', 'political_leader']]
```

```python
# How many countries are there whoes political leader is unknown

int(df[df['political_leader'].isna()]['country'].count())
```

```python
# How many countries have Republic in there full name 

# int(df['country_long'].str.contains('Republic').sum())
# or

count = 0
def check(txt):
    global count 
    if 'republic' in txt.lower():
        count +=1
    return count 

df['country_long'].apply(check)
print(count)
```

```python
# Which  country in africian region has highest population 

africa_df = df[df['continent'] == 'Africa']
africa_df.loc[africa_df['population'].idxmax()][['country', 'population']]

```

```python

# or

african_countries = df[df['region'].isin(['Northern Africa', 'Middle Africa', 'Western Africa', 'Southern Africa', 'Eastern Africa' ])]
african_countries.loc[african_countries['population'].idxmax(), ['country', 'population']]
```

```python
# or 
africa_df = df[df['continent'] == 'Africa']
africa_df[ africa_df['population'] == africa_df['population'].max()][['country', 'population']]
```
